```python
import openpyxl
from datetime import datetime

# ワークブックとシートの作成
workbook = openpyxl.Workbook()
sheet = workbook.active

# 会社名などの固定データの設定
company_name = '株式会社ABC'
address = '〒101-0022 東京都千代田区神田練塀町300'
tel_fax = 'TEL:03-1234-5678 FAX:03-1234-5678'
person_name = '担当者名:鈴木一郎 様'
number = '0001'

# 日付の設定
today = datetime.today()

# 日付のtodayオブジェクトを、特定のフォーマットで文字列に変換する
date = today.strftime('%Y/%m/%d')

# 商品データの設定
header = ['商品名', '数量', '単価', '金額']
data = [
    ['商品A', 2, 10000],
    ['商品B', 1, 15000],
]

# 固定データと日付の書き込み
sheet['B2'] = '請求書'
sheet['B4'] = company_name
sheet['B5'] = address
sheet['B6'] = tel_fax
sheet['B7'] = person_name
sheet['F4'] = 'No.'
sheet['G4'] = number
sheet['F5'] = '日付'
sheet['G5'] = date

# 商品ヘッダーの書き込み
sheet.cell(row=10, column=2, value=header[0])
sheet.cell(row=10, column=3, value=header[1])
sheet.cell(row=10, column=4, value=header[2])
sheet.cell(row=10, column=5, value=header[3])

# 商品データを書き込みつつ金額を合計
start_row = 11
total = 0
for i, datum in enumerate(data):
    amount = datum[1]*datum[2]
    sheet.cell(row=i+start_row, column=2, value=datum[0])
    sheet.cell(row=i+start_row, column=3, value=datum[1])
    sheet.cell(row=i+start_row, column=4, value=datum[2])
    sheet.cell(row=i+start_row, column=5, value=amount)
    total += amount

# 合計等の書き込み
last_row_of_item = sheet.max_row
sheet.cell(row=last_row_of_item+1, column=5, value=total)
sheet.cell(row=last_row_of_item+3, column=2, value='小計')
sheet.cell(row=last_row_of_item+3, column=5, value=total)
sheet.cell(row=last_row_of_item+4, column=2, value='消費税')
sheet.cell(row=last_row_of_item+4, column=5, value=total*0.1)
sheet.cell(row=last_row_of_item+5, column=2, value='合計')
sheet.cell(row=last_row_of_item+5, column=5, value=total*1.1)

# ワークブックの保存

# 日付のtodayオブジェクトを、特定のフォーマットで文字列に変換する
backup_filename = f"請求書_{today.strftime('%Y%m%d')}.xlsx"

workbook.save(backup_filename)

```
