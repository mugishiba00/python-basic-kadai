```python
class Human:

  def __init__(self,name,age):

    self.name = name

    self.age = age

  def check_adult(self):

    if self.age >= 20:
      print(self.name + ' is adult')

    else:
      print(self.name + ' is not adult')

humans = [Human('ichirou',19),Human('jirou',20),Human('saburou',21)]

for human in humans:
  human.check_adult()
```
